**Natural Language Processing (NLP)**

 NLP is a branch of artificial intelligence (AI) and computational linguistics that gives computers the ability to understand, interpret, generate, and manipulate human language (such as English, Spanish, or Hindi) in both text and speech forms.

Goal:

Bridge the gap between human communication and computer understanding by converting unstructured text or voice data into structured formats machine algorithms can process.

Common Applications:

Machine translation (e.g., Google Translate), sentiment analysis, chatbots, voice assistants (e.g., Alexa, Siri), text summarization, and spam detection.

Core Techniques

Tokenization:

Breaking text into individual words, terms, or symbols.

Stemming & Lemmatization:

Reducing words to their root or base forms (e.g., "running" to "run").

Parts-of-Speech (POS) Tagging:

Marking words as nouns, verbs, adjectives, etc., based on context.

Parsing & Named Entity Recognition (NER):

Identifying real-world entities like people, organizations, locations, or dates within text

**Step1:- Sentence Tokenization**

```python
text = "Backgammon is one of the oldest known games. It's history can be traced back 500 years to archeological discoveries in the Middle East.It is a two player game where each game has 15 checkers which moves between twenty-four points according to the roll of two dice."
```

```python
sentence = nltk.sent_tokenize(text)
for i in sentence:
  print(i)
  print()
```

**word tokenization**

```python
for w in sentence:
  words = nltk.word_tokenize(w)
  print(words)
  print()
```

**remove the stop words and get the root words**

there is 2 process --> to get the root words

1.stemming :- crude approach because it deletes the end letters from words --> ing,ed,er,es,.........etc

2.Lemmatization :- rule based approach --> where based on parts of speech we get root words

```python
from nltk.corpus import stopwords
import string
```

```python
from nltk.stem import PorterStemmer,WordNetLemmatizer
stopwords = set(stopwords.words("english"))
print(stopwords)
```

```python
#initialize the stemming and lemmatizer
lemmatizer = WordNetLemmatizer()
stemmer = PorterStemmer()

for w in sentence:
    words = nltk.word_tokenize(w)
 #remove the punctuations
    words = [word for word in words if word not in string.punctuation]
    print(words)
    #get the root words from above words
    lemmatized_words = [lemmatizer.lemmatize(word) for word in words]
    stemming_words = [stemmer.stem(word) for word in words]
    print("original words:",w)
    print("lemmatize_words:",lemmatized_words)
    print("stemming_words:",stemming_words)
    print()
```
